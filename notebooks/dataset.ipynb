```python
import numpy as np
import ipywidgets as widgets
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader


from src.dataset import GalacticBinariesDataset
from config import train_dataset_path, val_dataset_path
```

```python
train_dataset = GalacticBinariesDataset(train_dataset_path, max_K=10, max_samples=10_000)
```

```python
batch_size = 16
train_loader = DataLoader(train_dataset, batch_size=batch_size, num_workers=0, shuffle=True)
```

```python
summed_waveforms, k = next(iter(train_loader))
print("Summed waveforms shape:", summed_waveforms.shape)
print("K shape:", k.shape)
```

```python
waves = summed_waveforms.numpy() if hasattr(summed_waveforms, 'numpy') else summed_waveforms
k_vals = k.numpy() if hasattr(k, 'numpy') else k

batch_size, num_channels, time_steps = waves.shape

def explore_waveforms(batch_idx, channel_idx):
    plt.figure(figsize=(12, 5), facecolor='white')
    
    plt.plot(waves[batch_idx, channel_idx, :], color='#2B4A8E', linewidth=1)
    
    plt.title(f"Summed Waveforms (Sample {batch_idx} | Channel {channel_idx} | K: {k_vals[batch_idx]})", fontsize=14)
    plt.xlabel("Frequency Bin")
    plt.ylabel("Strain")
    plt.grid(True, linestyle='-', alpha=0.4)
    plt.xlim(0, time_steps)
    
    plt.tight_layout()
    plt.show()

batch_slider = widgets.IntSlider(
    min=0, max=batch_size-1, step=1, value=0, 
    description='Batch (n):', continuous_update=False
)

channel_slider = widgets.IntSlider(
    min=0, max=num_channels-1, step=1, value=0, 
    description='Channel (D1):', continuous_update=False
)

_ = widgets.interact(explore_waveforms, batch_idx=batch_slider, channel_idx=channel_slider)
```
